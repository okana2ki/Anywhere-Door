問１：リスト a の要素を合計するPythonプログラムを下のコードセルに書きなさい。また、いろいろなリストに対してそのプログラムが正しく動作することを確認しなさい。ヒント：空リストもリストの一種ですが、空リストでも正しく動作しますか？

```python
def sum_list_elements(a):
    return sum(a)

# さまざまなリストで動作を確認

# 通常のリスト
list1 = [1, 2, 3, 4, 5]
print(f"リスト {list1} の合計: {sum_list_elements(list1)}")

# 負の数を含むリスト
list2 = [-1, 0, 10, -5]
print(f"リスト {list2} の合計: {sum_list_elements(list2)}")

# 小数を含むリスト
list3 = [1.5, 2.5, 3.0]
print(f"リスト {list3} の合計: {sum_list_elements(list3)}")

# 空リスト
empty_list = []
print(f"空リスト {empty_list} の合計: {sum_list_elements(empty_list)}")

# 1つの要素のみのリスト
single_element_list = [42]
print(f"リスト {single_element_list} の合計: {sum_list_elements(single_element_list)}")
```

上級者用のおまけ問題 問１´：問１において、リストの要素が文字列だった場合、どうなりますか？文字列でも数値でも動作するようにできますか？文字列でも動作するとは、リスト`['a', 'b', 'c']`の要素を足し合わせると`'abc'`になるといった具合です。ヒント：文字列と数値で場合分けする必要があると思います。場合分け不要の洗練された方法があるかもしれませんが。

この「問１´」のような、**普通見ないような問題は、生成AIは苦手で、エラーが出るプログラムや、仕様通りの答が出ないプログラムを生成する可能性が高くなります**。それを経験してもらうための問題として「問１´」を作りました。**このような場合に、望むプログラムを作るためには、AIとどのようにやりとりをしたらいいか**、ぜひ、試行錯誤してみて下さい。→2年前は、こういう状態だったのですが、1年前、確認すると、**この程度の問題であれば、正しいプログラムを生成してくれる**（ことが多い？）ようでした。**AIは急速に進歩**していますからね。特に、プログラム生成は、生成AIが実用的に役立つ用途（生成AI向きの用途）なので、各社が競っていて、性能向上も極めて急速です。今、どれくらいの実力か、**モデルを切り替えながら**、いろいろ試して下さい。

```python
def sum_or_concatenate_list(a):
    if not a:
        # 空のリストの場合、sum()は0を返し、''.join()は''を返す。
        # どちらを返すか明確にするため、ここでは0をデフォルトとして返します。
        return 0

    try:
        # まず数値の合計を試みる
        return sum(a)
    except TypeError:
        # 合計が失敗した場合（文字列が含まれる場合など）、文字列の結合を試みる
        try:
            # Python 3では、map(str, a)は不要で、要素が既に文字列であればjoinできる
            # ここでは、リスト内の全要素が文字列であることを期待
            return "".join(a)
        except TypeError:
            # 結合も失敗した場合（数値と文字列が混在している場合など）、エラーを出す
            raise ValueError("リストの要素が数値と文字列の混合型であるため、合計または結合できません。")

# テストケース

print("--- 数値リストのテスト ---")
list_numbers1 = [1, 2, 3, 4, 5]
print(f"リスト {list_numbers1} の結果: {sum_or_concatenate_list(list_numbers1)}")

list_numbers2 = [-1.5, 0, 10]
print(f"リスト {list_numbers2} の結果: {sum_or_concatenate_list(list_numbers2)}")

print("\n--- 文字列リストのテスト ---")
list_strings1 = ['a', 'b', 'c']
print(f"リスト {list_strings1} の結果: {sum_or_concatenate_list(list_strings1)}")

list_strings2 = ["Hello", " ", "World"]
print(f"リスト {list_strings2} の結果: {sum_or_concatenate_list(list_strings2)}")

print("\n--- 空リストのテスト ---")
empty_list_test = []
print(f"空リスト {empty_list_test} の結果: {sum_or_concatenate_list(empty_list_test)}")

print("\n--- 1要素リストのテスト ---")
single_num_list = [42]
print(f"リスト {single_num_list} の結果: {sum_or_concatenate_list(single_num_list)}")
single_str_list = ['X']
print(f"リスト {single_str_list} の結果: {sum_or_concatenate_list(single_str_list)}")

print("\n--- 混合型リストのテスト (エラーが予想される) ---")
mixed_list1 = [1, 'b', 3]
try:
    print(f"リスト {mixed_list1} の結果: {sum_or_concatenate_list(mixed_list1)}")
except ValueError as e:
    print(f"エラー: {e}")

mixed_list2 = ['a', 2, 'c']
try:
    print(f"リスト {mixed_list2} の結果: {sum_or_concatenate_list(mixed_list2)}")
except ValueError as e:
    print(f"エラー: {e}")
```

問２：リスト a の要素を小さい順にソートするPythonプログラムを下のコードセルに書きなさい。また、いろいろなリストに対してそのプログラムが正しく動作することを確認しなさい。

参考：(組み込み関数`sorted` の定義は[公式ドキュメント](https://docs.python.org/ja/3/library/functions.html#sorted)を参照)

```python
def sort_list_ascending(a):
    return sorted(a)

# さまざまなリストで動作を確認

# 通常の数値リスト
list1 = [5, 2, 8, 1, 9]
print(f"元のリスト: {list1}, ソート後: {sort_list_ascending(list1)}")

# 負の数を含むリスト
list2 = [-3, 0, -10, 7]
print(f"元のリスト: {list2}, ソート後: {sort_list_ascending(list2)}")

# 小数を含むリスト
list3 = [3.14, 1.618, 2.718]
print(f"元のリスト: {list3}, ソート後: {sort_list_ascending(list3)}")

# 文字列リスト
list4 = ['banana', 'apple', 'cherry']
print(f"元のリスト: {list4}, ソート後: {sort_list_ascending(list4)}")

# 空リスト
empty_list = []
print(f"元のリスト: {empty_list}, ソート後: {sort_list_ascending(empty_list)}")

# 1つの要素のみのリスト
single_element_list = [42]
print(f"元のリスト: {single_element_list}, ソート後: {sort_list_ascending(single_element_list)}")

# 重複する要素を含むリスト
list5 = [3, 1, 4, 1, 5, 9, 2, 6]
print(f"元のリスト: {list5}, ソート後: {sort_list_ascending(list5)}")
```

問３：リスト a の要素を大きい順にソートするPythonプログラムを下のコードセルに書きなさい。また、いろいろなリストに対してそのプログラムが正しく動作することを確認しなさい。

問４：リスト a の先頭の要素を１つ削除して、末尾に35という値の要素を付け加えるPythonプログラムを下のコードセルに書きなさい。また、いろいろなリストに対してそのプログラムが正しく動作することを確認しなさい。

```python
def modify_list_elements(a):
    # リストが空でない場合のみ先頭要素を削除
    if a:
        a.pop(0)
    # 末尾に35を追加
    a.append(35)
    return a

# さまざまなリストで動作を確認

# 通常の数値リスト
list1 = [1, 2, 3, 4, 5]
print(f"元のリスト: {list1}, 変更後: {modify_list_elements(list1)}")

# 単一要素のリスト
list2 = [100]
print(f"元のリスト: {list2}, 変更後: {modify_list_elements(list2)}")

# 空リスト
list3 = []
print(f"元のリスト: {list3}, 変更後: {modify_list_elements(list3)}")

# 文字列を含むリスト
list4 = ['apple', 'banana', 'cherry']
print(f"元のリスト: {list4}, 変更後: {modify_list_elements(list4)}")

# 複数の要素を持つリスト
list5 = [10, 20, 30, 40, 50]
print(f"元のリスト: {list5}, 変更後: {modify_list_elements(list5)}")
```

問７：下のコードセルで、Geminiにコード作成を依頼して下さい。依頼内容は次の通り：「sklearnのdigitsデータを読み込んで、主成分分析で2次元に落として散布図を書いて。元画像と復元画像の比較もして。グラフ中の日本語出力は、japanize-matplotlibを使って。」

```python
# 必要なライブラリをインポート
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
import japanize_matplotlib # 日本語表示のためにインポート

# digitsデータを読み込む
digits = load_digits()
X = digits.data
y = digits.target

# 主成分分析 (PCA) を2次元に設定
pca_2d = PCA(n_components=2)
X_pca_2d = pca_2d.fit_transform(X)

# 2次元に落とした散布図をプロット
plt.figure(figsize=(10, 8))
scatter = plt.scatter(X_pca_2d[:, 0], X_pca_2d[:, 1], c=y, cmap='tab10', s=50, alpha=0.8)
plt.colorbar(scatter, ticks=range(10), label='数字')
plt.xlabel('第一主成分')
plt.ylabel('第二主成分')
plt.title('DigitsデータセットのPCA (2次元)')
plt.grid(True)
plt.show()

# 主成分の累積寄与率のグラフを描画
# 全ての主成分を使ってPCAを実行
pca_full = PCA()
pca_full.fit(X)

# 寄与率と累積寄与率を計算
explained_variance_ratio = pca_full.explained_variance_ratio_
cumulative_explained_variance = np.cumsum(explained_variance_ratio)

# 累積寄与率のグラフをプロット
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(cumulative_explained_variance) + 1), cumulative_explained_variance, marker='o')
plt.xlabel('主成分の数')
plt.ylabel('累積寄与率')
plt.title('主成分分析の累積寄与率')
plt.grid(True)
plt.show()

# 各数字の代表的なサンプルインデックスを見つける
sample_indices_per_digit = {}
for i in range(len(y)):
    if y[i] not in sample_indices_per_digit:
        sample_indices_per_digit[y[i]] = i
    if len(sample_indices_per_digit) == 10: # 全ての数字が見つかったら終了
        break

# 数字の昇順でソート (0, 1, ..., 9)
sorted_sample_items = sorted(sample_indices_per_digit.items())

# 試すPCA成分数のリスト
n_components_list = [2, 4, 8, 16, 32, 64]
num_components_to_compare = len(n_components_list)

# 各数字のサンプルについて復元画像を比較表示
for digit_label, sample_idx in sorted_sample_items:
    original_image = X[sample_idx].reshape(8, 8)
    original_label = y[sample_idx]

    # サブプロットの数を決定 (元画像 + 各成分数の復元画像)
    num_plots_in_row = 1 + num_components_to_compare

    fig, axes = plt.subplots(1, num_plots_in_row, figsize=(2 * num_plots_in_row, 3))
    fig.suptitle(f'数字 {original_label} のサンプル画像 (インデックス: {sample_idx}) のPCA復元比較', fontsize=16)

    # 元画像を表示
    axes[0].imshow(original_image, cmap='gray')
    axes[0].set_title('元画像')
    axes[0].set_xticks([])
    axes[0].set_yticks([])

    # 各成分数での復元画像を表示
    for j, n_components in enumerate(n_components_list):
        pca_recon = PCA(n_components=n_components)
        # fit_transform は全体データ X に対して行う
        X_pca_transformed = pca_recon.fit_transform(X)
        # inverse_transform は PCA 空間から元の空間に戻す
        X_reconstructed_all = pca_recon.inverse_transform(X_pca_transformed)

        recon_image = X_reconstructed_all[sample_idx].reshape(8, 8)
        axes[j + 1].imshow(recon_image, cmap='gray')
        axes[j + 1].set_title(f'{n_components}成分')
        axes[j + 1].set_xticks([])
        axes[j + 1].set_yticks([])

    plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # タイトルと重ならないように調整
    plt.show()
```

- 第3主成分までの3次元の散布図にする
- （Colabではできないが）ターミナルから、または、VS Code等で下記のコードを実行すると、マウスで回転やズームができる。

```python
# 必要なライブラリをインポート
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
import japanize_matplotlib # 日本語表示のためにインポート

# digitsデータを読み込む
digits = load_digits()
X = digits.data
y = digits.target

# 主成分分析 (PCA) を3次元に設定
pca_3d = PCA(n_components=3)
X_pca_3d = pca_3d.fit_transform(X)

# 3次元に落とした散布図をプロット
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
scatter = ax.scatter(X_pca_3d[:, 0], X_pca_3d[:, 1], X_pca_3d[:, 2], c=y, cmap='tab10', s=50, alpha=0.8)
plt.colorbar(scatter, ticks=range(10), label='数字')
ax.set_xlabel('第一主成分')
ax.set_ylabel('第二主成分')
ax.set_zlabel('第三主成分')
ax.set_title('DigitsデータセットのPCA (3次元)')
plt.show()
```
